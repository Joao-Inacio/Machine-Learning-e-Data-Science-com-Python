```python
import os
import pickle
from pathlib import Path

import numpy as np

DRIVE_DIRECTORY = Path.cwd().parent / "data"
```

```python
with open(os.path.join(DRIVE_DIRECTORY, "credit.pkl"), "rb") as f:
    X_credit_treinamento, y_credit_treinamento, X_credit_teste, y_credit_teste = (
        pickle.load(f)
    )
```

```python
X_credit = np.concatenate((X_credit_treinamento, X_credit_teste), axis=0)
y_credit = np.concatenate((y_credit_treinamento, y_credit_teste), axis=0)
X_credit, y_credit
```

```python
arvore = pickle.load(open('../0.9-avaliacaode-algoritmos-de-classificacao/Avore.sav', 'rb'))
svm = pickle.load(open('../0.9-avaliacaode-algoritmos-de-classificacao/SVM.sav',  'rb'))
floresta = pickle.load(open('../0.9-avaliacaode-algoritmos-de-classificacao/floresta.sav',  'rb'))
```

```python
novo_registro = X_credit[199]
novo_registro = novo_registro.reshape(1, -1)
novo_registro.shape
```

```python
resposta_arvores = arvore.predict(novo_registro)
resposta_svm = svm.predict(novo_registro)
resposta_floresta = floresta.predict(novo_registro)
```

```python
resposta_arvores[0], resposta_floresta[0], resposta_svm[0]
```

```python
paga = 0
nao_paga = 0
if resposta_arvores[0] == 1:
    nao_paga += 1
else:
    paga +=1
if resposta_floresta[0] == 1:
    nao_paga += 1
else:
    paga +=1
if resposta_svm[0] == 1:
    nao_paga += 1
else:
    paga +=1
if paga > nao_paga:
    print('Cliente pagará o empréstimo')
elif paga == nao_paga:
    print('Empate')
else:
    print('Cliente não pagará o empréstimo')
```
