# **Regressão**

```python
import os
from pathlib import Path

import numpy as np
import pandas as pd
import plotly.express as px
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.tree import DecisionTreeRegressor

DRIVE_DIRECTORY = Path.cwd().parent / "data"
```

## Regressão Polinomial

### Base plano de saúde

```python
base_plano_saude2 = pd.read_csv(os.path.join(DRIVE_DIRECTORY, "plano_saude2.csv"))
base_plano_saude2
```

```python
X_planos_saude2 = base_plano_saude2.iloc[:, 0:1].values
y_planos_saude2 = base_plano_saude2.iloc[:, 1].values
X_planos_saude2
```

```python
y_planos_saude2
```

```python
poly = PolynomialFeatures(degree=2)
X_planos_saude2_poly = poly.fit_transform(X_planos_saude2)
X_planos_saude2.shape
```

```python
X_planos_saude2_poly.shape
```

```python
regressor_saude_polinomial = LinearRegression()
regressor_saude_polinomial.fit(X_planos_saude2_poly, y_planos_saude2)
```

```python
regressor_saude_polinomial.intercept_
```

```python
regressor_saude_polinomial.coef_
```

```python
novo = [[40]]
```

```python
novo = poly.transform(novo)
novo
```

```python
regressor_saude_polinomial.predict(novo)
```

```python
previsoes = regressor_saude_polinomial.predict(X_planos_saude2_poly)
previsoes
```

```python
grafico = px.scatter(x=X_planos_saude2[:, 0], y=y_planos_saude2)
grafico.add_scatter(x=X_planos_saude2[:, 0], y=previsoes, name='Regressão')
grafico.show()
```

### Base preço das casas

```python
base_casas = pd.read_csv(os.path.join(DRIVE_DIRECTORY, 'house_prices.csv'))
base_casas
```

```python
X_casas = base_casas.iloc[:, 3:19].values
X_casas
```

```python
y_casas = base_casas.iloc[:, 2].values
y_casas
```

```python
X_casas_train, X_casas_test, y_casas_train, y_casas_test = train_test_split(
    X_casas, y_casas, test_size=0.3, random_state=0
)
```

```python
X_casas_train.shape, X_casas_test.shape
```

```python
poly = PolynomialFeatures(degree=2)
X_casas_train_poly = poly.fit_transform(X_casas_train)
X_casas_test_poly = poly.transform(X_casas_test)
```

```python
X_casas_train_poly.shape, X_casas_train_poly.shape
```

```python
regressor_casas_poly = LinearRegression()
regressor_casas_poly.fit(X_casas_train_poly, y_casas_train)
```

```python
regressor_casas_poly.score(X_casas_train_poly, y_casas_train)
```

```python
regressor_casas_poly.score(X_casas_test_poly, y_casas_test)
```

```python
previsoes = regressor_casas_poly.predict(X_casas_test_poly)
previsoes
```

```python
mean_absolute_error(y_casas_test, previsoes)
```

## Árvores de decisão

### Base plano saúde

```python
regressor_arvore_saude = DecisionTreeRegressor()
regressor_arvore_saude.fit(X_planos_saude2, y_planos_saude2)
```

```python
previsoes = regressor_arvore_saude.predict(X_planos_saude2)
previsoes
```

```python
regressor_arvore_saude.score(X_planos_saude2, y_planos_saude2)
```

```python
grafico = px.scatter(x=X_planos_saude2.ravel(), y=y_planos_saude2)
grafico.add_scatter(x=X_planos_saude2.ravel(), y=previsoes, name='Regressão')
grafico.show()
```

```python
X_teste_arvore = np.arange(X_planos_saude2.min(), X_planos_saude2.max(), 0.1)
X_teste_arvore
```

```python
X_teste_arvore = X_teste_arvore.reshape(-1, 1)
X_teste_arvore
```

```python
grafico = px.scatter(x=X_planos_saude2.ravel(), y=y_planos_saude2)
grafico.add_scatter(
    x=X_teste_arvore.ravel(),
    y=regressor_arvore_saude.predict(X_teste_arvore),
    name='Regressão')
grafico.show()
```

### Base preço das casas

```python
X_casas_train.shape
```

```python
regressor_arvore_casas = DecisionTreeRegressor()
regressor_arvore_casas.fit(X_casas_train, y_casas_train)
```

```python
regressor_arvore_casas.score(X_casas_train, y_casas_train)
```

```python
regressor_arvore_casas.score(X_casas_test, y_casas_test)
```

```python
previsoes = regressor_arvore_casas.predict(X_casas_test)
```

```python
previsoes
```

```python
mean_absolute_error(y_casas_test, previsoes)
```
