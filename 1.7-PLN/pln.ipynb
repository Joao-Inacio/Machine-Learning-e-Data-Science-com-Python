# **Processamento de Linguagem Natural com Python**

```python
import urllib.request

import bs4 as bs
import nltk
import spacy
from IPython.core.display import HTML
from spacy.matcher import PhraseMatcher

nltk.download('rslp')
```

## Marcação POS

```python
pln = spacy.load("pt_core_news_sm")
pln
```

```python
documento = pln('Estou aprendendo processamento de linguagem natural, curso em Curitiba')
type(documento)
```

```python
for token in documento:
    print(token.text, token.pos_)
```

## Lematizaçção e stemização

```python
for token in documento:
    print(token.text)
    print(token.lemma_)
```

```python
doc = pln('Encontrei encontraram encontrarão encontrariam cursando curso cursei')
[token.lemma_ for token in doc]
```

```python
stemmer = nltk.stem.RSLPStemmer()
stemmer.stem('aprender')
```

```python
for token in documento:
    print(token.text, token.lemma_, stemmer.stem(token.text))
```

## Carregamento dos textos

```python
url = 'https://pt.wikipedia.org/wiki/Intelig%C3%AAncia_artificial'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'}

req = urllib.request.Request(url, headers=headers)

try:
    # Tenta fazer a requisição
    resposta = urllib.request.urlopen(req)
    dados = resposta.read()
    # print(dados)  # Mostra o conteúdo da página
except urllib.error.HTTPError as e:
    print(f'Erro HTTP: {e.code}')
except urllib.error.URLError as e:
    print(f'Erro de URL: {e.reason}')
```

```python
dados
```

```python
dados_html = bs.BeautifulSoup(dados, 'lxml')
dados_html
```

```python
paragrafos = dados_html.find_all('p')
len(paragrafos)
```

```python
conteudo = ''
for p in paragrafos:
    conteudo += p.text

conteudo
```

```python
conteudo = conteudo.lower()
conteudo
```

## Buscas em textos com spaCy

```python
pln
```

```python
string = 'turing'
token_pesquisa = pln(string)
```

```python
matcher = PhraseMatcher(pln.vocab)
matcher.add('SEARCH', None, token_pesquisa)
```

```python
doc = pln(conteudo)
matches = matcher(doc)
matches
```

```python
texto = ''
numero_palavras = 50
doc = pln(conteudo)
matches = matcher(doc)

display(HTML(f'<h1>{string.upper()}</h1>'))
display(HTML(f"""<p><strong>Resultados encontrados:</strong> {len(matches)}</p>"""))
for i in matches:
    inicio = i[1] - numero_palavras
    inicio = max(inicio, 0)
    texto += str(doc[inicio:i[2] + numero_palavras]).replace(string, f"<mark>{string}</mark>")
    texto += "<br/><br/>"
display(HTML(f"""... {texto} ..."""))
```
