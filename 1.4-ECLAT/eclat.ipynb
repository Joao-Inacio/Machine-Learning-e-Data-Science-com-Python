# **ECLAT**

```python
import os
from pathlib import Path

import pandas as pd
from pyECLAT import ECLAT

DRIVE_DIRECTORY = Path.cwd().parent / "data"
```

```python
df_mercado1 = pd.read_csv(os.path.join(DRIVE_DIRECTORY, "mercado.csv"), header=None)
df_mercado1
```

```python
eclat = ECLAT(data=df_mercado1)
eclat.df_bin
```

```python
eclat.uniq_
```

```python
indices, suporte = eclat.fit(min_support=0.3, min_combination=1, max_combination=3)
indices
```
