# **Regras de associação**

## Algoritmo Apriori

```python
import os
from pathlib import Path

import pandas as pd
from apyori import apriori

DRIVE_DIRECTORY = Path.cwd().parent / "data"
```

## Base de dados mercado 1

```python
df_mercado1 = pd.read_csv(os.path.join(DRIVE_DIRECTORY, "mercado.csv"), header=None)
df_mercado1
```

```python
transacoes = []
for i in range(len(df_mercado1)):
    transacoes.append([str(df_mercado1.values[i, j]) for j in range(df_mercado1.shape[1])])
transacoes
```

```python
regras = apriori(transacoes, min_support=0.3, min_confidence=0.8, min_lift=2)
resultado = list(regras)
resultado
```

```python
len(resultado)
```

```python
resultado[0]
```

```python
r = resultado[2][2]
r
```

```python
se = []
entao = []
suporte = []
confianca = []
lift = []

for resultados in resultado:
    s = resultados[1]
    result_rules = resultados[2]
    for result_rule in result_rules:
        a = list(result_rule[0])
        b = list(result_rule[1])
        c = result_rule[2]
        l = result_rule[3]

        se.append(a)
        entao.append(b)
        suporte.append(s)
        confianca.append(c)
        lift.append(l)
```

```python
rules__df = pd.DataFrame({
    'Se': se,
    'Então': entao,
    'Suporte': suporte,
    'Confianca': confianca,
    'Lift': lift
})
rules__df.sort_values(by='Lift', ascending=False)
```
